```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
## sentences
sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'I love learning Python',
    'Python is easy to learn',
    'I am learning machine learning',
    'I am learning deep learning',
    'I understand the meaning of words',
    'I enjoy working with data',
    'Data science is interesting',
    'Artificial intelligence is powerful',
    'Your videos are good'
]
```

```python
sent
```

```python
## define the vocabulary size
voc_size = 10000
```

```python
## one hot representation of the sentences
one_hot_repr=[one_hot(words, voc_size) for words in sent]
one_hot_repr
## here you can see that each word is represented a unique index number. The index number is generated based on the hash function. The same word will always have the same index number.
```

```python
## word embedding representation
from tensorflow.keras.layers import Embedding
from tensorflow.keras.utils import pad_sequences ## see all sentences are not of same lenth some are of 4 words and some are of 5 words. So we need to make all sentences of same length. For that we will use pad_sequences function.pad_sequences add zeros from front or back of the sentence to make all sentences of same length. By default it adds zeros from front of the sentence.
from tensorflow.keras.models import Sequential
import numpy as np
```

```python
sent_length = 8
embedded_docs = pad_sequences(one_hot_repr, padding='pre', maxlen=sent_length)
print(embedded_docs)
```

```python
## feature representation of the sentences
dim =10
```

```python
model = Sequential()
model.add(Embedding(voc_size, dim, input_length=sent_length))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
